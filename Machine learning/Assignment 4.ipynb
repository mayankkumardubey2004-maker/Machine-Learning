```python
import pandas as pd
```

```python
import numpy as np
```

```python
df = pd.read_csv("Housing.csv")
```

```python
df.info()
```

```python
# question 1

result = df.groupby("bedrooms")["price"].mean().sort_values(ascending=False)
print(result)
```

```python
# question 2
result = df.groupby("stories")["area"].sum().sort_values(ascending=False).head(2)
print(result)
```

```python
# question 3

grouped = df.groupby("hotwaterheating")

max_price = grouped["price"].max()
print(max_price)

result = df.loc[df.groupby("hotwaterheating")["price"].idxmax()]
print(result)
```

```python
# question 4

df["PricePerSqFt"] = df["price"] / df["area"]

result = df.groupby("stories")["PricePerSqFt"].mean()
print(result)
```

```python
# question 5

result = df.groupby("parking").size().sort_values()
print(result)
```

```python
# question 6

result = df.groupby("bathrooms")["area"].agg(["min", "max", "mean"])
print(result)
```

```python
# question 7

grouped = df.groupby("bedrooms")

print(grouped.first())
print(grouped.last())
print(grouped.sum())
print(grouped.mean(numeric_only=True))
print(grouped.count())
print(grouped.max())
print(grouped.min())
print(grouped["price"].describe())
```

```python
# question 8

grouped = df.groupby("furnishingstatus")

for name, group in grouped:
    print("Group Name:", name)
    print(group)
    print("-" * 50)
```
