```python
import seaborn as sns
```

```python
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
tips= sns.load_dataset("tips")
```

```python
tips.head()
```

## Categorical scatterplot

```python
sns.stripplot(data=tips, x='day', y='total_bill')
```

```python
sns.stripplot(data=tips, x='day', y='total_bill',hue='sex')
```

```python
sns.stripplot(data=tips, x='day', y='total_bill',jitter=False)
```

## 2.swarmplot -> shows the distribution of data more clear

```python
sns.swarmplot(data=tips,x='day',y='total_bill')
```

```python
sns.catplot(data=tips,x='day',y='total_bill',kind='swarm')
```

## Categorical Distribution plot
   1.Boxplot :- used for visualization

   2.Violinplot

```python
sns.boxplot(data=tips,x='sex',y='total_bill')
```

```python
sns.boxplot(data=tips,x='day',y='total_bill')
```

```python
sns.boxplot(data=tips,x='sex',y='total_bill',hue='sex')
```

## Violinplot

```python
sns.violinplot(data=tips,x='day',y='total_bill')
```

```python
sns.violinplot(data=tips,x='sex',y='total_bill',hue='sex')
```

## Categorical estimate plots
barplot


pointplot

countplot

1.barplot()
-> The upper black lines are known as errorbar

```python
sns.barplot(data=tips,x='sex',y='total_bill')
```

```python
sns.barplot(data=tips,x='sex',y='total_bill',hue='sex')
```

2. pointplot()
-> It just remove the bars from bar plot and connect the top most points of the bars.

```python
sns.pointplot(data=tips,x='day',y='total_bill')
```

```python
sns.pointplot(data=tips,x='day',y='total_bill',hue='sex')
```

```python
sns.pointplot(data=tips,x='sex',y='total_bill',hue='sex')
```

## 3. countplot()

```python
sns.countplot(data=tips,x='day')
```

```python
sns.countplot(data=tips,x='day',hue='sex')
```
