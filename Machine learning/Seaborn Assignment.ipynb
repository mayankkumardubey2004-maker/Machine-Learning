```python
import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset("penguins")
```

```python
df.sample(10)
```

## 1. Relationship Between Body Mass and Flipper Length Across Species

```python
sns.scatterplot(data=df, x="body_mass_g", y="flipper_length_mm", hue="species")

sns.relplot(data=df, x="body_mass_g", y="flipper_length_mm", hue="species")
```

## 2.Trend of Body Mass by Island

```python
sns.lineplot(data=df, x="island", y="body_mass_g", errorbar="sd")
sns.relplot(data=df, x="island", y="body_mass_g", kind="line", errorbar="sd")
```

## 3.Comparison of Body Mass Across Species

```python
sns.barplot(data=df,x='species',y='body_mass_g',hue='sex')
sns.catplot(data=df,x='species',y='body_mass_g',kind='bar',hue='sex')
```

## 4.Distribution of Flipper Length by Species and Gender

```python

# Violin plot
sns.violinplot(data=df, x="species", y="flipper_length_mm", hue="sex")


# Figure-level
sns.catplot(data=df, x="species", y="flipper_length_mm", hue="sex", kind="box")
```

## 5. Count of Penguins by Species and Island

```python
sns.countplot(data=df,x='species',hue='island')

# Figure-level
sns.catplot(data=df, x="species", hue="island", kind='count')
```

## 6.Pairwise Relationship Between Features

```python
sns.pairplot(data=df, hue="species", diag_kind="hist")
```

## 7.Body Mass Distribution Across Species

```python
# Axis-level
sns.histplot(data=df, x="body_mass_g", hue="species", kde=True)


# Figure-level
sns.displot(data=df, x="body_mass_g", hue="species", kde=True)
```

## 8.Joint Distribution of Flipper Length and Bill Length

```python
# Scatter
sns.jointplot(data=df, x="flipper_length_mm", y="bill_length_mm", kind="scatter")
# KDE
sns.jointplot(data=df, x="flipper_length_mm", y="bill_length_mm", kind="kde")
# Hex
sns.jointplot(data=df, x="flipper_length_mm", y="bill_length_mm", kind="hex")
```

## 9.Study the Relationship Between Body Mass and Island

```python
sns.jointplot(data=df, x="body_mass_g", y="flipper_length_mm", hue="island")
```

## 10.Explore the Relationship Between Body Mass and Bill Depth

```python
# Scatter
sns.jointplot(data=df, x="body_mass_g", y="bill_depth_mm", kind="scatter")
# KDE
sns.jointplot(data=df, x="body_mass_g", y="bill_depth_mm", kind="kde")
```
